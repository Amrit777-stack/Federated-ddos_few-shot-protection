```python
import os
import pandas as pd
import numpy as np

data_path = "../data/raw/CSV-01-12/01-12"
processed_path = "../data/processed"

files = [f for f in os.listdir(data_path) if f.endswith(".csv")]

print("CSV files found:")
for f in files:
    print(f)

print("\nTotal files:", len(files))
```

```python
samples = []
target_per_class = 1000

target_labels = {
    "BENIGN",
    "DrDoS_DNS",
    "DrDoS_LDAP",
    "DrDoS_MSSQL",
    "DrDoS_NetBIOS",
    "DrDoS_NTP",
    "DrDoS_SNMP",
    "DrDoS_SSDP",
    "DrDoS_UDP",
    "Syn",
    "TFTP",
    "UDP-lag",
    "WebDDoS"
}

collected = {label: 0 for label in target_labels}

for file in files:
    path = os.path.join(data_path, file)

    for chunk in pd.read_csv(path, chunksize=100000):
        chunk.columns = chunk.columns.str.strip()

        for label in target_labels:
            if collected[label] < target_per_class:
                needed = target_per_class - collected[label]

                selected = chunk[chunk["Label"] == label].head(needed)

                if len(selected) > 0:
                    samples.append(selected)
                    collected[label] += len(selected)

        if all(count >= target_per_class for count in collected.values()):
            break

    print(file, "processed")

    if all(count >= target_per_class for count in collected.values()):
        break

dataset = pd.concat(samples, ignore_index=True)

print("\nCollected samples:")
print(collected)

print("\nDataset shape:", dataset.shape)
```
